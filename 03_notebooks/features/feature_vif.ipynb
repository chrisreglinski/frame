```python
import pandas as pd
import numpy as np

abt = pd.read_parquet('C:/Users/twluc/frame/01_data/03_abt/major/abt.parquet')

reps1 = [
    'hmt_season_goals_diff_avg',
    'awt_season_goals_diff_avg',
    'mrkt_home_away_impl_diff',
    'mrkt_entropy',
    'hmt_season_goals_agst_avg',
    'hmt_season_goals_total_avg',
    'awt_season_goals_agst_avg',
    'awt_season_goals_total_avg',
]

reps2 = [
    'hmt_season_goals_diff_avg',
    'awt_season_goals_diff_avg',
    'mrkt_home_away_impl_diff',
    'mrkt_entropy',
    'hmt_season_goals_agst_avg',
    'hmt_season_goals_for_avg',
    'awt_season_goals_agst_avg',
    'awt_season_goals_for_avg',
]
```

```python
def vif(cols):
    # VIF = diagonal of the inverse correlation matrix.
    # >5-10 means multicollinear; huge/inf means a variable is a (near-)exact
    # linear combination of the others (something pairwise |corr| cannot show).
    corr = abt[cols].dropna().corr()
    return pd.Series(np.diag(np.linalg.inv(corr.values)), index=cols).sort_values(ascending=False).round(1)
```

```python
vif(reps1)
```

```python
# goals_agst = (goals_total - goals_diff) / 2, so per side {diff, agst, total} are only 2 dims.
# drop goals_agst on each side and the collinearity is gone.
vif([c for c in reps1 if 'goals_agst' not in c])
```

```python
vif(reps2)
```

```python
# goals_agst = (goals_total - goals_diff) / 2, so per side {diff, agst, total} are only 2 dims.
# drop goals_agst on each side and the collinearity is gone.
vif([c for c in reps2 if 'goals_diff' not in c])
```
