```python
# simple linear regression on the representative variables. 
# checking also if predicted t_goals_diff predicts draws well. 
# result: not great.

from sklearn.linear_model import LinearRegression
from sklearn.metrics import r2_score

import numpy as np

features = [
    'hmt_season_impl_win_avg',
    'mrkt_home_impl',
    'hmt_season_impl_draw_avg',
]
target = 't_goals_diff'

data = abt[features + [target, 'season']].dropna()
train = data[data['season'] != '2526']
test  = data[data['season'] == '2526']

lr = LinearRegression()
lr.fit(train[features], train[target])

pred_train = lr.predict(train[features])
pred_test  = np.round(lr.predict(test[features]))

print(f'R² train: {r2_score(train[target], pred_train):.4f}')
print(f'R² test:  {r2_score(test[target],  pred_test):.4f}')

pd.Series(lr.coef_, index=features).sort_values()
```

```python
test
```

```python
fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for ax, (label, y_true, y_pred) in zip(axes, [
    ('train', train[target], pred_train),
    ('test',  test[target],  pred_test),
]):
    ax.scatter(y_true, y_pred, alpha=0.3, s=15)
    lims = [min(y_true.min(), y_pred.min()) - 0.5,
            max(y_true.max(), y_pred.max()) + 0.5]
    ax.plot(lims, lims, 'r--', linewidth=1)
    ax.set_xlabel('actual')
    ax.set_ylabel('predicted')
    ax.set_title(f'{label}  R²={r2_score(y_true, y_pred):.4f}')

plt.tight_layout()
plt.show()
```

```python
test_full = (
    test[[target] + features]
    .assign(pred=pred_test.round(2))
    .sort_values('pred')
    .reset_index(drop=True)
)
test_full

```

```python
mask = pred_test.round() == 0
idx  = test.index[mask]

impl  = abt.loc[idx, 'mrkt_draw_impl']
y_drw = abt.loc[idx, 't_draw_flg'].astype(int)

# stawiamy na wszystkie — proba ustawiamy powyżej implied żeby portfolio_roi zakwalifikował każdy mecz
proba = impl + 0.1

profit = (y_drw * (1/impl - 1) + (1 - y_drw) * (-1)).sum()
roi    = profit / len(idx)
print(f'n={len(idx)}  profit={profit:.2f}  roi={roi:.3f}')
```
