```python
import frame_for_notebooks as frame
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

ts = frame.build_match_team_stats()
mt = frame.build_match_target()

df = ts.merge(mt[['match_id','t_goals_diff','t_goals_total','t_draw_flg']], on='match_id')
df['n_min'] = df[['home_game_number','away_game_number']].min(axis=1)

# ── 1. STABILIZACJA KORELACJI FEATURE → TARGET ───────────────────────
df['gdiff_diff'] = df.home_season_goals_diff_avg - df.away_season_goals_diff_avg
df['openness']   = df.home_season_goals_total_avg + df.away_season_goals_total_avg

pairs = [('gdiff_diff', 'home_goals_diff'),
         ('openness',   'total_goals')]

ns = range(2, 31)
fig, axes = plt.subplots(1, 2, figsize=(13, 4))
for ax, (feat, target) in zip(axes, pairs):
    corrs = [df[df.n_min == n][[feat, target]].corr().iloc[0, 1] for n in ns]
    counts = [(df.n_min == n).sum() for n in ns]
    ax.plot(ns, corrs, marker='o')
    ax.set_title(f'corr({feat}, {target}) vs game number')
    ax.set_xlabel('min(home_n, away_n)')
    ax.axhline(0, color='gray', lw=0.5)
plt.tight_layout()
plt.show()
```
