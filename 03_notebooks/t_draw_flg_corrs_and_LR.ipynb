```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.calibration import CalibrationDisplay

features = [
    'mrkt_draw_impl',
    'hmt_season_impl_draw_avg',
    #'b365_away_odds',
    #'hmt_season_impl_win_avg',
    #'hmt_season_shots_on_target_for_avg',
    #'hmt_season_goals_diff_avg',
]
target = 't_draw_flg'

data = abt[features + [target, 'season']].dropna()
train = data[data['season'] != '2526']
test  = data[data['season'] == '2526']

lr = LogisticRegression(max_iter=1000)
lr.fit(train[features], train[target])

proba_train = lr.predict_proba(train[features])[:, 1]
proba_test  = lr.predict_proba(test[features])[:, 1]

auc_train = roc_auc_score(train[target], proba_train)
auc_test  = roc_auc_score(test[target],  proba_test)
print(f'AUC  train={auc_train:.4f}  test={auc_test:.4f}')

# calibration
fig, axes = plt.subplots(1, 2, figsize=(12, 4))
for ax, (label, y, p) in zip(axes, [
    ('train', train[target], proba_train),
    ('test',  test[target],  proba_test),
]):
    CalibrationDisplay.from_predictions(y, p, n_bins=10, ax=ax, name='LR')
    ax.set_title(f'calibration — {label}  (AUC={roc_auc_score(y, p):.4f})')
plt.tight_layout()
plt.show()

# coefs
pd.Series(lr.coef_[0], index=features).sort_values()
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.calibration import CalibrationDisplay

features = [
    'mrkt_draw_impl',
    'hmt_season_impl_draw_avg',
    #'b365_away_odds',
    'hmt_season_impl_win_avg',
    'hmt_season_shots_on_target_for_avg',
    #'hmt_season_goals_diff_avg',
]
target = 't_draw_flg'

data = abt[features + [target, 'season']].dropna()

# --- leave-one-season-out CV (all 4 seasons) ---
cv_rows = []
for fold_season in sorted(data['season'].unique()):
    cv_tr = data[data['season'] != fold_season]
    cv_te = data[data['season'] == fold_season]

    lr = LogisticRegression(max_iter=1000)
    lr.fit(cv_tr[features], cv_tr[target])

    proba = pd.Series(lr.predict_proba(cv_te[features])[:, 1], index=cv_te.index)
    auc   = roc_auc_score(cv_te[target], proba)
    res   = portfolio_roi(proba, cv_te['mrkt_draw_impl'], cv_te[target].astype(int), buffer=0.01)
    cv_rows.append({'season': fold_season, 'auc': round(auc, 4), **res})

cv_df = pd.DataFrame(cv_rows).set_index('season')
cv_df.loc['mean'] = cv_df.mean()
display(cv_df)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import roc_auc_score
from sklearn.preprocessing import PolynomialFeatures
from sklearn.pipeline import make_pipeline

features = [
    'mrkt_draw_impl',
    #'hmt_season_impl_draw_avg',
    #'b365_away_odds',
    'hmt_season_impl_win_avg',
    'hmt_season_shots_on_target_for_avg',
    #'hmt_season_goals_diff_avg',
]
target = 't_draw_flg'

data = abt[features + [target, 'season']].dropna()

cv_rows = []
for fold_season in sorted(data['season'].unique()):
    cv_tr = data[data['season'] != fold_season]
    cv_te = data[data['season'] == fold_season]

    model = make_pipeline(
        PolynomialFeatures(degree=2, include_bias=False),
        LogisticRegression(max_iter=1000, C=0.1),
    )
    model.fit(cv_tr[features], cv_tr[target])

    proba = pd.Series(model.predict_proba(cv_te[features])[:, 1], index=cv_te.index)
    auc   = roc_auc_score(cv_te[target], proba)
    res   = portfolio_roi(proba, cv_te['mrkt_draw_impl'], cv_te[target].astype(int), buffer=0.01)
    cv_rows.append({'season': fold_season, 'auc': round(auc, 4), **res})

cv_df = pd.DataFrame(cv_rows).set_index('season')
cv_df.loc['mean'] = cv_df.mean()
display(cv_df)
```
